```python
from PIL import Image, ImageDraw

# Crie uma nova imagem
largura, altura = 500, 300
imagem = Image.new('RGB', (largura, altura), 'lightblue')
desenho = ImageDraw.Draw(imagem)

# Desenhe o corpo do pássaro
desenho.ellipse((150, 100, 250, 200), fill='brown')

# Desenhe as asas
desenho.polygon([(150, 150), (50, 50), (150, 100)], fill='white')
desenho.polygon([(250, 150), (350, 50), (250, 100)], fill='white')

# Desenhe o bico
desenho.polygon([(200, 100), (250, 80), (200, 120)], fill='yellow')

# Salve a imagem
imagem.save('passaro_voando.png')
```
